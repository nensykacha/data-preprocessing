#### 2. Describe how to plan a Data Science Project, listing all steps.


## Question 2: Describe How to Plan a Data Science Project

Planning a Data Science project means deciding the steps needed to solve a problem using data. Proper planning helps us complete the project in an organized way and get useful results.

### Steps of a Data Science Project

1. **Understand the Business Problem:**

   First, we need to understand the problem that we want to solve. We should clearly decide the main goal of the project and what results we want to get.

2. **Data Collection:**

   In this step, we collect the required data from different sources, such as CSV files, databases, APIs, and websites. The collected data is used for further analysis.

3. **Data Understanding and Exploration (EDA):**

   After collecting the data, we study it to understand its structure and values. We use charts, graphs, and statistical methods to find patterns, trends, and unusual values.

4. **Data Cleaning and Preprocessing:**

   In this step, we clean the data by handling missing values, removing duplicate records, and correcting errors. We also prepare the data in a suitable format for analysis and model building.

5. **Feature Selection and Engineering:**

   Feature selection means choosing the important columns needed for the project. Feature engineering means creating or changing columns to make the data more useful for building a Machine Learning model.

6. **Model Building and Evaluation:**

   In this step, we choose a suitable Machine Learning algorithm and train the model using the prepared data. We then check its performance using measures such as Accuracy, Precision, Recall, and F1-Score.

7. **Deployment and Maintenance:**

   After testing the model, we make it available for real use through an application, API, or dashboard. We also check its performance regularly and update it when needed.

## Question 4: Explain Tensors and Give Examples Using NumPy

### What is a Tensor?

A tensor is a data structure used to store numbers in one or more dimensions. It is a general form of a scalar, vector, and matrix.

Tensors are widely used in Machine Learning and Deep Learning. They help store input data, model values, and calculation results. Libraries such as TensorFlow and PyTorch use tensors to perform different calculations.

### Types of Tensors Based on Dimensions

1. **Scalar (0-D Tensor):**

   A scalar is a single numerical value. It has zero dimensions.

   Example: `45` or `4.5`

2. **Vector (1-D Tensor):**

   A vector is a list of numerical values. It has one dimension.

   Example: `[25, 30, 45]`, which can represent the ages of customers.

3. **Matrix (2-D Tensor):**

   A matrix is a collection of values arranged in rows and columns. It has two dimensions.

   Example: A table containing customer ages, incomes, and purchase details.

4. **3-D and Higher-Dimensional Tensors:**

   These tensors store data in three or more dimensions. They are useful for working with images, videos, and time-based data.

   Example: A colour image can be represented using height, width, and colour channels.

### Python Code Examples Using NumPy

NumPy is a Python library used to work with arrays and perform mathematical calculations. We can use NumPy to create different types of tensors and check their dimensions.

### 1. Scalar (0-D Tensor)

A scalar stores a single numerical value.

```python
import numpy as np

scalar_tensor = np.array(45)

print("0-D Tensor (Scalar):", scalar_tensor)
print("Dimensions:", scalar_tensor.ndim)
```

### 2. Vector (1-D Tensor)

A vector stores a list of numerical values.

```python
vector_tensor = np.array([22, 35, 40, 28, 52])

print("1-D Tensor (Vector):", vector_tensor)
print("Dimensions:", vector_tensor.ndim)
```

### 3. Matrix (2-D Tensor)

A matrix stores values in rows and columns.

```python
matrix_tensor = np.array([
    [25, 50000, 5],
    [32, 72000, 8],
    [47, 120000, 12]
])

print("2-D Tensor (Matrix):\n", matrix_tensor)
print("Dimensions:", matrix_tensor.ndim)
print("Shape:", matrix_tensor.shape)
```

### 4. 3-D Tensor

A 3-D tensor stores data in three dimensions.

```python
tensor_3d = np.array([
    [[25, 50000], [32, 72000]],
    [[47, 120000], [19, 30000]]
])

print("3-D Tensor:\n", tensor_3d)
print("Dimensions:", tensor_3d.ndim)
print("Shape:", tensor_3d.shape)
```

### Explanation of the Code

- `import numpy as np` is used to import the NumPy library.
- `np.array()` is used to create an array.
- `ndim` tells us the number of dimensions of an array.
- `shape` tells us the size of the array in each dimension.
- A scalar has zero dimensions, a vector has one dimension, a matrix has two dimensions, and the last example has three dimensions.

### Conclusion

Tensors are important in Machine Learning and Deep Learning because they help store and process numerical data. NumPy makes it easy to create arrays and understand their dimensions and shapes.

### Part C: Data Understanding & Cleaning

#### 6. Perform Initial Exploration

First, we will explore the dataset using .head(), .info(), and .describe() methods, and check for missing values and duplicates.

````python
import pandas as pd

# Load the dataset
df = pd.read_csv('customer_purchases.csv')

print("--- 1. First 5 Rows (.head()) ---")
display(df.head())

print("\n--- 2. Dataset Information (.info()) ---")
df.info()

print("\n--- 3. Statistical Summary (.describe()) ---")
display(df.describe())

print("\n--- 4. Check for Missing Values ---")
print(df.isnull().sum())

print("\n--- 5. Check for Duplicate Rows ---")
print("Duplicate count:", df.duplicated().sum())
````

#### 7. Apply Data Cleaning

Next, we perform data cleaning by handling missing values (if any), correcting data types, and removing irrelevant columns.

````python
# -------------------------------------------------------------
# A. Handle Missing Data (Imputation or Removal)
# -------------------------------------------------------------
# If there are missing values in numerical columns, fill with median
if df['Income'].isnull().sum() > 0:
    df['Income'].fillna(df['Income'].median(), inplace=True)

# If there are missing values in categorical columns, fill with mode
if df['Gender'].isnull().sum() > 0:
    df['Gender'].fillna(df['Gender'].mode()[0], inplace=True)

# Alternatively, you can drop rows with missing values if needed:
# df.dropna(inplace=True)


# -------------------------------------------------------------
# B. Correct Inconsistent Data Types (Example)
# -------------------------------------------------------------
# Convert 'CustomerID' to string or object if it's treated purely as an identifier
df['CustomerID'] = df['CustomerID'].astype(str)

# If you had a date column, you would convert it like this:
# df['SignupDate'] = pd.to_datetime(df['SignupDate'])


# -------------------------------------------------------------
# C. Drop Irrelevant Columns
# -------------------------------------------------------------
# If there are columns that are not needed for analysis, drop them
# For example, dropping an unneeded column (if present)
df.drop(columns=['UnnecessaryColumnName'], inplace=True, errors='ignore')

print("Data Cleaning Completed Successfully!")
print("Cleaned Data Shape:", df.shape)
display(df.head())
````

### Part D: Exploratory Data Analysis (EDA)

First, make sure to import the required visualization libraries at the top of your EDA section:

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set graph design style
sns.set(style="whitegrid")
df = pd.read_csv('customer_purchases.csv')
````

#### 8. Univariate Analysis (Distribution plots of Age, Income, and Purchases)

````python
plt.figure(figsize=(15, 4))

# 1. Age Distribution Plot
plt.subplot(1, 3, 1)
sns.histplot(df['Age'], kde=True, color='blue')
plt.title('Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')

# 2. Income Distribution Plot
plt.subplot(1, 3, 2)
sns.histplot(df['Income'], kde=True, color='green')
plt.title('Income Distribution')
plt.xlabel('Income')
plt.ylabel('Count')

# 3. Purchases Distribution Plot
plt.subplot(1, 3, 3)
sns.histplot(df['Purchases'], kde=True, color='orange')
plt.title('Purchases Distribution')
plt.xlabel('Purchases')
plt.ylabel('Count')

plt.tight_layout()
plt.show()

# Short Interpretation: These plots show the frequency range of customer ages, how income is distributed among 
# customers, and how many purchases are typically made.
````

#### 9. Bivariate Analysis (Relationship between two variables)

Bivariate analysis explores the relationship between two different variables.

#### A. Relationship between Gender & Purchases

````python
plt.figure(figsize=(6, 4))
sns.boxplot(x='Gender', y='Purchases', data=df, palette='Set2')
plt.title('Gender vs Purchases')
plt.xlabel('Gender')
plt.ylabel('Purchases')
plt.show()

#Short Interpretation: This box plot compares the purchasing patterns between male and female customers.
````

#### B. Relationship between Income & Churn

````python
plt.figure(figsize=(6, 4))
sns.boxplot(x='Churn', y='Income', data=df, palette='Pastel1')
plt.title('Income vs Churn (0 = Stayed, 1 = Left)')
plt.xlabel('Churn Status')
plt.ylabel('Income')
plt.show()

#Short Interpretation: This helps us analyze whether customers with lower or higher incomes are more likely to churn (leave the company).
````

#### 10. Multivariate Analysis (Multiple variables together)

Multivariate analysis helps us see correlations and interactions between three or more variables.

#### A. Correlation Heatmap of all numerical variables

````python

plt.figure(figsize=(8, 6))
numeric_df = df.select_dtypes(include=[np.number])
correlation_matrix = numeric_df.corr()

sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f", linewidths=0.5)
plt.title('Correlation Heatmap')
plt.show()

# Short Interpretation: The heatmap shows how strongly numerical columns (like Age, Income, and Purchases) are related to each other. Values close to $1$ mean strong positive correlation.
````

#### B. Pair plots to identify feature interactions

````python
sns.pairplot(df, hue='Churn', palette='husl', diag_kind='kde')
plt.show()

# Short Interpretation: The pair plot creates a grid of scatter plots and distribution curves, color-coded by whether a customer churned or stayed, allowing us to spot visual patterns easily.
````
